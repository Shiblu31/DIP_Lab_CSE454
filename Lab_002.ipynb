```python
# ==========================================
# Image Resizing using OpenCV
# ==========================================

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Read image
image_path = list(uploaded.keys())[0]
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Original image size
print("Original Shape:", image.shape)

# Resize Image (Width = 600, Height = 400)
resized = cv2.resize(image, (600, 400))

print("Resized Shape:", resized.shape)

# Display Images
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(resized)
plt.title("Resized Image")
plt.axis("off")

plt.show()
```

```python
# ==========================================
# Image Cropping using OpenCV (Automatic Center Crop)
# ==========================================

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# ------------------------------------------
# Upload Image
# ------------------------------------------
uploaded = files.upload()

# Get uploaded file name
image_path = list(uploaded.keys())[0]

# ------------------------------------------
# Read Image
# ------------------------------------------
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# ------------------------------------------
# Get Image Size
# ------------------------------------------
height, width = image.shape[:2]

print("Original Image Shape :", image.shape)
print("Height :", height)
print("Width  :", width)

# ------------------------------------------
# Crop the Center 50% of the Image
# ------------------------------------------

crop_width = width // 2
crop_height = height // 2

# Starting coordinates
start_x = (width - crop_width) // 2
start_y = (height - crop_height) // 2

# Ending coordinates
end_x = start_x + crop_width
end_y = start_y + crop_height

# Crop Image
cropped = image[start_y:end_y, start_x:end_x]

print("\nCropping Coordinates")
print(f"Start Point : ({start_x}, {start_y})")
print(f"End Point   : ({end_x}, {end_y})")

print("\nCropped Image Shape :", cropped.shape)

# ------------------------------------------
# Display Images
# ------------------------------------------
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cropped)
plt.title("Cropped Image")
plt.axis("off")

plt.show()
```

```python
# ==========================================
# Image Rotation using OpenCV
# ==========================================

import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Read image
image_path = list(uploaded.keys())[0]
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

height, width = image.shape[:2]

# Center of Image
center = (width // 2, height // 2)

# Rotation Matrix
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1)

# Rotate Image
rotated = cv2.warpAffine(image,
                         rotation_matrix,
                         (width, height))

# Display Images
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(rotated)
plt.title("Rotated Image (45°)")
plt.axis("off")

plt.show()
```
