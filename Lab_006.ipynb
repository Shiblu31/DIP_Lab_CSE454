```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# 1. Brightness Transformation
brightness_value = 50
bright_image = cv2.convertScaleAbs(
    image, alpha=1, beta=brightness_value
)

# 2. Contrast Transformation
contrast_value = 1.5
contrast_image = cv2.convertScaleAbs(
    image, alpha=contrast_value, beta=0
)

# 3. Negative Transformation
negative_image = 255 - image

# Display all images
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(bright_image)
plt.title("Brightness Transformation")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(contrast_image)
plt.title("Contrast Transformation")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(negative_image)
plt.title("Negative Transformation")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Brightness value
brightness_value = 50

# Increase brightness
bright_image = np.clip(
    image.astype(np.int16) + brightness_value,
    0, 255
).astype(np.uint8)

# Display original and bright images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(bright_image)
plt.title("Bright Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Contrast value
contrast_value = 1.5

# Apply contrast transformation
contrast_image = np.clip(
    contrast_value * image.astype(np.float32),
    0, 255
).astype(np.uint8)

# Display original and contrast images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(contrast_image)
plt.title("Contrast Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
