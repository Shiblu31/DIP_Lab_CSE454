```python

# Lab 01: Read and Display Image in Colab

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# ------------------------------------------
# Upload Image
# ------------------------------------------
uploaded = files.upload()

# Get uploaded file name
image_path = list(uploaded.keys())[0]

# ------------------------------------------
# Read Image using OpenCV
# ------------------------------------------
image = cv2.imread(image_path)

# Check whether image is loaded
if image is None:
    print("Error! Image could not be loaded.")
else:
    print("Image loaded successfully.")

# ------------------------------------------
# Image Information
# ------------------------------------------
print("\nImage Shape :", image.shape)
print("Image Height:", image.shape[0])
print("Image Width :", image.shape[1])
print("Channels    :", image.shape[2])

# ------------------------------------------
# Convert BGR to RGB
# ------------------------------------------
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# ------------------------------------------
# Display Original Image
# ------------------------------------------
plt.figure(figsize=(8,6))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()

# ------------------------------------------
# Display Grayscale Image
# ------------------------------------------
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(8,6))
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.show()

# ------------------------------------------
# Save Processed Image
# ------------------------------------------
cv2.imwrite("grayscale_output.jpg", gray)

print("\nGrayscale image saved successfully!")

# ------------------------------------------
# Pixel Information
# ------------------------------------------
print("\nPixel at (100,100):")

if image.shape[0] > 100 and image.shape[1] > 100:
    print("BGR :", image[100,100])
    print("Gray:", gray[100,100])
else:
    print("Image is smaller than 101×101 pixels.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Read image
image_path = list(uploaded.keys())[0]
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(image)
plt.axis("off")
plt.title("Original Image")
plt.show()

# Print image dimensions
print("Image Shape:", image.shape)

# Convert image to pixel array
pixel_array = np.array(image)

# Display pixel values
print("\nPixel Matrix:")
print(pixel_array)

# Access a specific pixel
row = 100
col = 150

print(f"\nPixel value at ({row}, {col}) = {pixel_array[row, col]}")
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Image information
print("Shape :", image.shape)
print("Height:", image.shape[0])
print("Width :", image.shape[1])
print("Channels:", image.shape[2])
print("Data Type:", image.dtype)

# Access pixel
print("\nPixel at (100,100):", image[100,100])

# Modify pixel
image[100,100] = [0,0,255]

# Draw shapes
cv2.circle(image,(250,250),50,(255,0,0),3)
cv2.rectangle(image,(50,50),(200,200),(0,255,0),3)
cv2.line(image,(0,0),(400,400),(255,255,0),4)

# Add text
cv2.putText(image,
            "DIP Lab",
            (50,450),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (255,0,0),
            2)

# Convert to RGB
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display image
plt.figure(figsize=(8,6))
plt.imshow(rgb)
plt.axis("off")
plt.title("Basic Image Operations")
plt.show()

# Save image
cv2.imwrite("output.jpg", image)
print("Image saved successfully.")
```
