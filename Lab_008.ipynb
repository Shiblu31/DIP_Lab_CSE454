```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image in grayscale
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# --------------------------------
# 1. Mean Filtering
# --------------------------------
mean_filter = cv2.blur(image, (3, 3))

# --------------------------------
# 2. Weighted Average Filtering
# --------------------------------
kernel = np.array([
    [1, 2, 1],
    [2, 4, 2],
    [1, 2, 1]
], dtype=np.float32) / 16

weighted_filter = cv2.filter2D(
    image, -1, kernel
)

# --------------------------------
# 3. Max Filtering
# --------------------------------
max_filter = cv2.dilate(
    image, np.ones((3, 3), np.uint8)
)

# --------------------------------
# 4. Min Filtering
# --------------------------------
min_filter = cv2.erode(
    image, np.ones((3, 3), np.uint8)
)

# --------------------------------
# 5. Median Filtering
# --------------------------------
median_filter = cv2.medianBlur(image, 3)

# --------------------------------
# Display Results
# --------------------------------
plt.figure(figsize=(14, 10))

plt.subplot(2, 3, 1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(mean_filter, cmap='gray')
plt.title("Mean Filtering")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(weighted_filter, cmap='gray')
plt.title("Weighted Average Filtering")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(max_filter, cmap='gray')
plt.title("Max Filtering")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(min_filter, cmap='gray')
plt.title("Min Filtering")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(median_filter, cmap='gray')
plt.title("Median Filtering")
plt.axis("off")

plt.tight_layout()
plt.show()
```
