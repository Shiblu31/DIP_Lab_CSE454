```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get image name
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ------------------------------------------
# Digital Negative
# Formula: Negative = 255 - Pixel
# ------------------------------------------

negative = 255 - gray

# ------------------------------------------
# Display Images
# ------------------------------------------

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap='gray')
plt.title("Digital Negative")
plt.axis("off")

plt.show()

# ------------------------------------------
# Display Some Pixel Values
# ------------------------------------------

print("Original pixel value:", gray[50, 50])
print("Negative pixel value:", negative[50, 50])

# Save result
cv2.imwrite("digital_negative.jpg", negative)

print("\nDigital negative saved successfully.")
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get image name
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ------------------------------------------
# Simple Binary Thresholding
# ------------------------------------------

threshold_value = 127

ret, binary = cv2.threshold(
    gray,
    threshold_value,
    255,
    cv2.THRESH_BINARY
)

# ------------------------------------------
# Display Images
# ------------------------------------------

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binary, cmap='gray')
plt.title("Binary Thresholded Image")
plt.axis("off")

plt.show()

print("Threshold Value:", ret)

# Save result
cv2.imwrite("thresholded_image.jpg", binary)

print("Thresholded image saved successfully.")
```

```python
# ==========================================
# Image Intensity Clipping
# ==========================================

import cv2
import matplotlib.pyplot as plt
from google.colab import files
import numpy as np

# Upload image
uploaded = files.upload()

# Get image name
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# ------------------------------------------
# Set Clipping Range
# ------------------------------------------

lower_limit = 50
upper_limit = 200

# ------------------------------------------
# Apply Intensity Clipping
# ------------------------------------------

clipped = np.clip(
    gray,
    lower_limit,
    upper_limit
)

# Convert data type to uint8
clipped = clipped.astype(np.uint8)

# ------------------------------------------
# Display Images
# ------------------------------------------

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(clipped, cmap='gray')
plt.title("Clipped Image")
plt.axis("off")

plt.show()

# ------------------------------------------
# Display Pixel Example
# ------------------------------------------

print("Example Pixel Values")

print("Original pixel:", gray[50, 50])
print("Clipped pixel :", clipped[50, 50])

# Save result
cv2.imwrite("clipped_image.jpg", clipped)

print("\nClipped image saved successfully.")
```
